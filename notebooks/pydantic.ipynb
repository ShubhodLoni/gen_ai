```python
from pydantic import BaseModel, Field
```

```python
class User(BaseModel):
    name : str
    age : int
```

```python
user = User(name="shubhod",age=20)
user
```

field validation

```python
class Product(BaseModel):
    title : str = Field(max_length = 10) 
    des : str = Field(min_length = 20)
    price : int = Field(gt=0,le=100)
```

```python
product = Product(title="iphone 18",des="amazing performance phone is iphone",price=100)
product
```

multiple class validation

```python
class Order(BaseModel):
    id:str = Field(min_length=0,max_length=100)
    product:Product
```

```python
order = Order(id="sjdiag",product=product)
order
```
