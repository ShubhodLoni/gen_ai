#### Text Splitter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
texts = "Artificial Intelligence (AI) is a branch of computer science that enables machines to perform tasks that typically require human intelligence."\
"It is used in various applications such as virtual assistants, recommendation systems, and self-driving cars."
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=50, chunk_overlap=20)
chunks = splitter.split_text(texts)
```
